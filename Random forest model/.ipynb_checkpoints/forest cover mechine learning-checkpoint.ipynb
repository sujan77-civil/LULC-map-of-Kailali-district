```python
import ee
import geemap
```

```python
ee.Authenticate()
```

```python
ee.Initialize()
```

```python
# Create interactive map
Map = geemap.Map()
```

# =============================================================================
# SECTION 2: Load AOI and Boundaries
# =============================================================================

# Load landslide polygons and boundary

```python
aoi = ee.FeatureCollection('projects/primeval-rune-481716-s3/assets/polygon')
illamBdry = ee.FeatureCollection('projects/primeval-rune-481716-s3/assets/Illam_boundary')
```

# Center map and add layers 

```python
Map.centerObject(aoi, 13)
Map.addLayer(aoi, {'color': 'red'}, 'AOI')
Map.addLayer(illamBdry, {'color': 'blue'}, 'ILLAM_BDRY')
```

create map

```python
Map
```

# =============================================================================
# SECTION 3: Load and Process Sentinel-2 Data
# =============================================================================
# Load Sentinel-2 Harmonized Surface Reflectance

```python
s2 = ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
```

```python
# Filter for Post-Event Image
filtered_s2_post = (s2
                    .filterBounds(aoi)
                    .filterDate('2025-10-01', '2025-12-31')
                    .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 20)))

# Create median composite and clip to AOI
S2_Median_Post = filtered_s2_post.median().clip(aoi)

# Visualize RGB composite
visual_parameters = {
    'bands': ['B4', 'B3', 'B2'],
    'min': 0,
    'max': 3000
}

Map.addLayer(S2_Median_Post, visual_parameters, 'S2_Median_Post')
```

# =============================================================================
# SECTION 4: Load Pre-Event Sentinel-2 Data
# =============================================================================

```python
# Filter for Pre-Event Image
S2_Pre = (s2
          .filterBounds(aoi)
          .filterDate('2024-10-01', '2024-12-31')
          .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 20))
          .median()
          .clip(aoi))

print("Pre and Post event images loaded successfully")
```

# =============================================================================
# SECTION 5: Calculate and Visualize NDVI
# =============================================================================

```python
# Calculate NDVI
ndvi = S2_Median_Post.normalizedDifference(['B8', 'B4']).rename('NDVI')

# Define visualization parameters
ndvi_vis = {
    'min': 0,
    'max': 0.8,
    'palette': [
        '#FFFFFF', '#CE7E45', '#DF923D', '#F1B555', '#FCD163', '#99B718', '#74A901',
        '#66A000', '#529400', '#3E8601', '#207401', '#056201', '#004C00', '#023B01',
        '#012E01', '#011D01', '#011301'
    ]
}

Map.addLayer(ndvi, ndvi_vis, 'NDVI Vegetation Index')
```

# =============================================================================
# SECTION 6: Calculate and Visualize NBR
# =============================================================================

```python
# Calculate NBR (Normalized Burn Ratio)
nbr = S2_Median_Post.normalizedDifference(['B8', 'B12']).rename('NBR')

# Add NBR to map
Map.addLayer(nbr, {'min': -1, 'max': 1,
             'palette': ['red', 'white', 'green']}, 'NBR')

```

# =============================================================================
# SECTION 7: Calculate and Visualize NDMI
# =============================================================================

```python
# Calculate NDMI (Normalized Difference Moisture Index)
ndmi = S2_Median_Post.normalizedDifference(['B8', 'B11']).rename('NDMI')

# Define visualization
ndmi_vis = {
    'min': -0.2,
    'max': 0.8,
    'palette': ['#d73027', '#f46d43', '#fdae61', '#fee08b', '#d9ef8b',
                '#a6d96a', '#66bd63', '#1a9850', '#053061']
}

Map.addLayer(ndmi, ndmi_vis, 'Moisture Index (NDMI)')
```

# =============================================================================
# SECTION 8: Calculate and Visualize BSI (Bare Soil Index)
# =============================================================================

```python
# Calculate BSI
bsi = S2_Median_Post.expression(
    '((SWIR1 + RED) - (NIR + BLUE)) / ((SWIR1 + RED) + (NIR + BLUE))', {
        'SWIR1': S2_Median_Post.select('B11'),
        'RED': S2_Median_Post.select('B4'),
        'NIR': S2_Median_Post.select('B8'),
        'BLUE': S2_Median_Post.select('B2')
    }).rename('BSI')

# Define visualization
bsi_vis = {
    'min': -1,
    'max': 1,
    'palette': ['green', 'white', 'magenta']
}

Map.addLayer(bsi, bsi_vis, 'Bare Soil Index (BSI)')
```

# =============================================================================
# SECTION 9: Calculate and Visualize GLCM Texture
# =============================================================================

```python
# Select NIR band and convert to 8-bit
nir = S2_Median_Post.select('B8')
nir8Bit = nir.unitScale(0, 3000).multiply(255).toByte()

# Calculate GLCM texture
glcm = nir8Bit.glcmTexture(size=4)

# Select contrast metric
contrast = glcm.select('B8_contrast')
variance = glcm.select('B8_var')

Map.addLayer(contrast, {'min': 0, 'max': 1000}, 'Terrain Contrast (Texture)')
```

# =============================================================================
# SECTION 10: Calculate and Visualize dNDVI (Vegetation Loss)
# =============================================================================

```python
# Calculate Pre-Event NDVI
ndvi_pre = S2_Pre.normalizedDifference(['B8', 'B4'])

# Calculate dNDVI (difference)
dNDVI = ndvi_pre.subtract(ndvi).rename('dNDVI')

# Visualize (red areas = vegetation loss)
Map.addLayer(dNDVI, {'min': 0.1, 'max': 0.5, 'palette': ['white', 'yellow', 'red']},
             'dNDVI (Vegetation Loss)')
```

# =============================================================================
# SECTION 11: Calculate and Visualize dNBR (Disturbance)
# =============================================================================

```python
# Calculate Pre-Event NBR
nbr_pre = S2_Pre.normalizedDifference(['B8', 'B12']).rename('NBR_pre')

# Calculate dNBR
dNBR = nbr_pre.subtract(nbr).rename('dNBR')

# Define visualization
dnbr_vis = {
    'min': -0.1,
    'max': 0.5,
    'palette': ['#7fb3d5', '#ffffff', '#f7dc6f', '#f39c12', '#d35400', '#922b21']
}

Map.addLayer(dNBR, dnbr_vis, 'dNBR (Disturbance Severity)')
```

# =============================================================================
# SECTION 12: Calculate and Visualize Slope
# =============================================================================

```python
# Load SRTM elevation data
srtm = ee.Image("USGS/SRTMGL1_003")

# Calculate slope in degrees
slope = ee.Terrain.slope(srtm).clip(aoi)

# Define visualization
slopeVis = {
    'min': 0,
    'max': 45,
    'palette': ['#228B22', '#FFFF00', '#FF0000']
}

# Add slope to map
Map.addLayer(slope, slopeVis, 'Slope (Degrees)')

# Optional: Highlight steep slopes (> 25 degrees)
steepSlopes = slope.gt(25)
Map.addLayer(steepSlopes.updateMask(steepSlopes), {'palette': ['black']},
             'Slopes > 25 Deg', False)
Map
```

# =============================================================================
# SECTION 13: Export Results (Optional)
# =============================================================================

```python
# Example: Export NDVI as GeoTIFF
task = ee.batch.Export.image.toDrive(
    image=ndvi,
    description='NDVI_Export',
    folder='GEE_Exports',
    region=aoi.geometry(),
    scale=10,
    crs='EPSG:4326',
    maxPixels=1e13
)

# Uncomment to start export
# task.start()
# print("Export task started")
```

# =============================================================================
# SECTION 14: Create Combined Multi-Band Image (Optional)
# =============================================================================

```python
# Combine all indices into one multi-band image
combined = (ee.Image.cat([
    ndvi,
    nbr,
    ndmi,
    bsi,
    contrast,
    dNDVI,
    dNBR,
    slope
]).clip(aoi))

print("All indices calculated successfully!")
print("Bands in combined image:", combined.bandNames().getInfo())
```
